```python
import requests
import pandas as pd
from google.colab import userdata, drive
import os

# 1. 구글 드라이브 연결 (마운트)
drive.mount('/content/drive')

# 2. 저장 경로 설정 (알려주신 경로로 지정)
# 내 드라이브의 기본 경로는 /content/drive/MyDrive/ 입니다.
save_path = '/content/drive/MyDrive/NotebookLM/문화 관광/경주'

# 해당 폴더가 없으면 자동으로 만듭니다.
if not os.path.exists(save_path):
    os.makedirs(save_path)
    print(f"📂 새 폴더를 생성했습니다: {save_path}")

# 3. 인증키 및 데이터 수집
service_key = userdata.get('MY_KEY')
url = 'http://apis.data.go.kr/5050000/cafeInfoService/getCafeInfo'
params = {'serviceKey': service_key, 'pageNo': '1', 'numOfRows': '100', 'type': 'json'}

try:
    response = requests.get(url, params=params)
    items = response.json().get('response', {}).get('body', {}).get('items', {}).get('item', [])

    if items:
        # 데이터 가공
        df = pd.DataFrame(items)
        final_df = df[['CON_TITLE', 'CON_ADDRESS', 'CON_PHONE', 'CON_SUMMARY', 'CON_CONTENT']].copy()
        final_df.columns = ['카페 이름', '주소', '전화번호', '운영 및 주차 정보', '카페 특징 상세']

        # 태그 및 특수문자 제거 정리
        final_df['운영 및 주차 정보'] = final_df['운영 및 주차 정보'].str.replace(r'<[^>]*>|&nbsp;', ' ', regex=True)
        final_df['카페 특징 상세'] = final_df['카페 특징 상세'].str.replace(r'<[^>]*>|&nbsp;', ' ', regex=True)

        # 4. 구글 드라이브에 직접 저장 (기존 파일이 있으면 덮어씁니다)
        file_full_path = os.path.join(save_path, 'gyeongju_cafe_complete.csv')
        final_df.to_csv(file_full_path, index=False, encoding='utf-8-sig')

        print(f"✅ 성공! 파일이 다음 경로에 저장되었습니다:")
        print(f"📍 {file_full_path}")
        display(final_df.head(3))
    else:
        print("❌ 데이터를 가져오지 못했습니다. 인증키를 확인해 주세요.")

except Exception as e:
    print(f"⚠️ 오류 발생: {e}")
```
